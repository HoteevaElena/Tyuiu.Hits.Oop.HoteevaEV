Создать базовый класс Task в C#, который будет представлять задачи внутри
проекта. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Task:

• Атрибуты: ID задачи (TaskId), Название задачи (TaskName), Приоритет
задачи (Priority).

• Методы:
o MarkAsComplete(): метод для отметки задачи как выполненной.
o GetTaskDetails(): метод для получения деталей задачи.
o ReassignTo(): метод для переназначения задачи другому члену
команды.

Требования к производным классам:
1. ДелегатскаяЗадача (DelegateTask): Должна содержать дополнительные
атрибуты, такие как Дата выполнения (DueDate).
Метод MarkAsComplete() должен быть переопределен для включения даты
выполнения в сообщение о завершении задачи.
2. КоманднаяЗадача (TeamTask): Должна содержать дополнительные атрибуты,
такие как Команда (TeamName). Метод ReassignTo() должен быть
переопределен для указания нового члена команды, которому будет
переназначена задача.
3. ИсследовательскаяЗадача (ResearchTask) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Исходные данные
(DataSource). Метод GetTaskDetails() должен быть переопределен для
отображения источников данных, используемых в задаче, вместе с другими
деталями задачи.

```
// Интерфейс для всех задач
public interface ITask
{
    void MarkAsComplete();
    string GetTaskDetails();
    void ReassignTo(string newAssignee);
    int CalculateEstimatedHours();
    bool ValidateTask();
    void SetReminder();
}

public class Task : ITask
{
    // Базовые атрибуты
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public int Priority { get; set; }
    public string Description { get; set; } // Новый атрибут
    public DateTime CreatedDate { get; set; } // Новый атрибут
    public string Category { get; set; } // Новый атрибут
    
    private bool isCompleted = false;
    protected string assignedTo; // Новый атрибут

    public Task(int taskId, string taskName, int priority, string description, string category)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        Description = description;
        Category = category;
        CreatedDate = DateTime.Now;
        assignedTo = "Не назначено";
    }

    // Метод для отметки задачи как выполненной
    public virtual void MarkAsComplete()
    {
        isCompleted = true;
        Console.WriteLine($"Задача {TaskName} отмечена как выполненная");
    }

    // Метод для получения деталей задачи
    public virtual string GetTaskDetails()
    {
        string status = isCompleted ? "Выполнена" : "В процессе";
        return $"ID: {TaskId}, Название: {TaskName}, Приоритет: {Priority}, " +
               $"Категория: {Category}, Статус: {status}, Назначено: {assignedTo}";
    }

    // Метод для переназначения задачи
    public virtual void ReassignTo(string newAssignee)
    {
        assignedTo = newAssignee;
        Console.WriteLine($"Задача {TaskName} переназначена на {newAssignee}");
    }

    // Новый метод для расчета предполагаемого времени
    public virtual int CalculateEstimatedHours()
    {
        return Priority * 2; // Базовый расчет на основе приоритета
    }

    // Новый метод для валидации задачи
    public virtual bool ValidateTask()
    {
        return !string.IsNullOrEmpty(TaskName) && 
               !string.IsNullOrEmpty(Description) && 
               Priority >= 1 && Priority <= 5;
    }

    // Новый метод для установки напоминания
    public virtual void SetReminder()
    {
        Console.WriteLine($"Напоминание установлено для задачи: {TaskName}");
    }

    // Новый метод для получения статуса выполнения
    public bool IsCompleted()
    {
        return isCompleted;
    }
}

public class DelegateTask : Task
{
    // Дополнительные атрибуты
    public DateTime DueDate { get; set; }
    public string DelegatedBy { get; set; } // Новый атрибут
    public string DelegationReason { get; set; } // Новый атрибут

    public DelegateTask(int taskId, string taskName, int priority, string description, 
                       string category, DateTime dueDate, string delegatedBy, string delegationReason) 
        : base(taskId, taskName, priority, description, category)
    {
        DueDate = dueDate;
        DelegatedBy = delegatedBy;
        DelegationReason = delegationReason;
    }

    // Переопределенный метод для отметки задачи как выполненной
    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Дата выполнения: {DueDate}, Делегировал: {DelegatedBy}");
    }

    // Переопределенный метод для получения деталей задачи
    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Срок: {DueDate.ToShortDateString()}, " +
               $"Делегировал: {DelegatedBy}, Причина: {DelegationReason}";
    }

    // Новый метод для проверки просрочки
    public bool IsOverdue()
    {
        return DateTime.Now > DueDate && !IsCompleted();
    }

    // Переопределенный метод расчета времени
    public override int CalculateEstimatedHours()
    {
        int baseHours = base.CalculateEstimatedHours();
        TimeSpan timeUntilDue = DueDate - DateTime.Now;
        return (int)(baseHours * (timeUntilDue.TotalDays < 7 ? 1.5 : 1.0));
    }
}

public class TeamTask : Task
{
    // Дополнительные атрибуты
    public string TeamName { get; set; }
    public int TeamSize { get; set; } // Новый атрибут
    public string TeamLead { get; set; } // Новый атрибут
    
    // Конструктор
    public TeamTask(int taskId, string taskName, int priority, string description, 
                   string category, string teamName, int teamSize, string teamLead) 
        : base(taskId, taskName, priority, description, category)
    {
        TeamName = teamName;
        TeamSize = teamSize;
        TeamLead = teamLead;
    }

    // Переопределенный метод для переназначения задачи
    public override void ReassignTo(string newAssignee)
    {
        base.ReassignTo(newAssignee);
        Console.WriteLine($"Новая команда: {TeamName}, Руководитель: {TeamLead}");
    }

    // Переопределенный метод для получения деталей задачи
    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Команда: {TeamName}, " +
               $"Размер команды: {TeamSize}, Руководитель: {TeamLead}";
    }

    // Новый метод для расчета нагрузки на команду
    public double CalculateWorkloadPerPerson()
    {
        int estimatedHours = CalculateEstimatedHours();
        return (double)estimatedHours / TeamSize;
    }

    // Переопределенный метод валидации
    public override bool ValidateTask()
    {
        return base.ValidateTask() && TeamSize > 0 && !string.IsNullOrEmpty(TeamLead);
    }
}

public class ResearchTask : Task
{
    // Дополнительные атрибуты
    public string DataSource { get; set; }
    public string ResearchMethodology { get; set; } // Новый атрибут
    public int RequiredSamples { get; set; } // Новый атрибут
    
    // Конструктор
    public ResearchTask(int taskId, string taskName, int priority, string description, 
                       string category, string dataSource, string researchMethodology, int requiredSamples) 
        : base(taskId, taskName, priority, description, category)
    {
        DataSource = dataSource;
        ResearchMethodology = researchMethodology;
        RequiredSamples = requiredSamples;
    }

    // Переопределенный метод для получения деталей задачи
    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Источник данных: {DataSource}, " +
               $"Методология: {ResearchMethodology}, Требуется образцов: {RequiredSamples}";
    }

    // Переопределенный метод расчета времени
    public override int CalculateEstimatedHours()
    {
        int baseHours = base.CalculateEstimatedHours();
        return baseHours + (RequiredSamples / 10); // Добавляем время на сбор образцов
    }

    // Новый метод для проверки источников данных
    public bool ValidateDataSource()
    {
        string[] validSources = { "Google Scholar", "PubMed", "IEEE", "ScienceDirect" };
        return validSources.Contains(DataSource);
    }
}

// Класс для управления зависимостями (простой контейнер)
public class TaskManager
{
    private readonly List<Task> _tasks;

    public TaskManager()
    {
        _tasks = new List<Task>();
    }

    public void AddTask(Task task)
    {
        if (task.ValidateTask())
        {
            _tasks.Add(task);
            Console.WriteLine($"Задача '{task.TaskName}' добавлена в менеджер");
        }
        else
        {
            Console.WriteLine($"Ошибка: Задача '{task.TaskName}' не прошла валидацию");
        }
    }

    public void DisplayAllTasks()
    {
        Console.WriteLine("\n=== ВСЕ ЗАДАЧИ В МЕНЕДЖЕРЕ ===");
        foreach (var task in _tasks)
        {
            Console.WriteLine(task.GetTaskDetails());
        }
    }

    public void MarkTaskComplete(int taskId)
    {
        var task = _tasks.FirstOrDefault(t => t.TaskId == taskId);
        if (task != null)
        {
            task.MarkAsComplete();
        }
    }
}

// Явная реализация интерфейса для специального типа задачи
public class UrgentTask : ITask
{
    public string TaskName { get; set; }
    public int EmergencyLevel { get; set; }

    // Явная реализация интерфейса
    void ITask.MarkAsComplete()
    {
        Console.WriteLine($"СРОЧНО: Задача {TaskName} завершена! Уровень срочности: {EmergencyLevel}");
    }

    string ITask.GetTaskDetails()
    {
        return $"СРОЧНАЯ ЗАДАЧА: {TaskName}, Уровень срочности: {EmergencyLevel}";
    }

    void ITask.ReassignTo(string newAssignee)
    {
        Console.WriteLine($"СРОЧНО: Задача {TaskName} переназначена на {newAssignee}");
    }

    int ITask.CalculateEstimatedHours()
    {
        return Math.Max(1, 6 - EmergencyLevel); // Чем срочнее, тем меньше времени
    }

    bool ITask.ValidateTask()
    {
        return !string.IsNullOrEmpty(TaskName) && EmergencyLevel >= 1 && EmergencyLevel <= 5;
    }

    void ITask.SetReminder()
    {
        Console.WriteLine($"СРОЧНОЕ напоминание для: {TaskName}");
    }
}

        // Создаем менеджер задач (управление зависимостями)
        var taskManager = new TaskManager();

        Console.WriteLine("1. Создание базовой задачи:");
        Task basicTask = new Task(1, "Базовая задача", 2, "Описание базовой задачи", "Общие");
        Console.WriteLine(basicTask.GetTaskDetails());
        taskManager.AddTask(basicTask);
        
        Console.WriteLine("\n2. Создание делегированной задачи:");
        DelegateTask delegateTask = new DelegateTask(2, "Делегированная задача", 1, 
            "Важная делегированная задача", "Администрирование", 
            DateTime.Now.AddDays(3), "Иван Иванов", "Высокий приоритет");
        Console.WriteLine(delegateTask.GetTaskDetails());
        taskManager.AddTask(delegateTask);

        Console.WriteLine("\n3. Создание командной задачи:");
        TeamTask teamTask = new TeamTask(3, "Командная задача", 3, 
            "Разработка нового функционала", "Разработка", 
            "Команда разработки", 5, "Петр Петров");
        Console.WriteLine(teamTask.GetTaskDetails());
        taskManager.AddTask(teamTask);

        Console.WriteLine("\n4. Создание исследовательской задачи:");
        ResearchTask researchTask = new ResearchTask(4, "Исследовательская задача", 2, 
            "Анализ рынка", "Исследования", 
            "Google Scholar", "Количественный анализ", 100);
        Console.WriteLine(researchTask.GetTaskDetails());
        taskManager.AddTask(researchTask);

        // Демонстрация новых методов
        Console.WriteLine("\n5. Демонстрация новых методов:");
        Console.WriteLine($"Расчет времени для базовой задачи: {basicTask.CalculateEstimatedHours()} часов");
        Console.WriteLine($"Расчет времени для командной задачи: {teamTask.CalculateEstimatedHours()} часов");
        Console.WriteLine($"Нагрузка на человека в команде: {teamTask.CalculateWorkloadPerPerson():F1} часов/чел");
        Console.WriteLine($"Проверка источника данных: {researchTask.ValidateDataSource()}");

        // Демонстрация явной реализации интерфейса
        Console.WriteLine("\n6. Явная реализация интерфейса:");
        UrgentTask urgentTask = new UrgentTask { TaskName = "Критический баг", EmergencyLevel = 5 };
        ITask urgentTaskInterface = urgentTask; // Используем через интерфейс
        
        Console.WriteLine(urgentTaskInterface.GetTaskDetails());
        urgentTaskInterface.SetReminder();
        Console.WriteLine($"Расчетное время: {urgentTaskInterface.CalculateEstimatedHours()} часов");

        // Отметка задач как выполненных
        Console.WriteLine("\n7. Отмечаем задачи как выполненные:");
        basicTask.MarkAsComplete();
        delegateTask.MarkAsComplete();

        // Переназначение задач
        Console.WriteLine("\n8. Переназначаем задачи:");
        basicTask.ReassignTo("Алексей Сидоров");
        teamTask.ReassignTo("Мария Козлова");

        // Показываем все задачи в менеджере
        taskManager.DisplayAllTasks();

        // Демонстрация работы менеджера
        Console.WriteLine("\n9. Работа с менеджером задач:");
        taskManager.MarkTaskComplete(3);
```
