```
public class Task
{
    // Базовые атрибуты
    protected int _TaskId;
    protected string _TaskName;
    protected int _Priority;
    private string _assignee;
    protected bool isCompleted = false;
    public string About {set; get;}

    public int TaskId {get {return _TaskId;} set
    {
        if(value > 0)
        {
            _TaskId = value;
        }
        else{
            throw new ArgumentException("ID задачи должен быть положительным числом");
        }
    }
    }

    public string TaskName {get {return _TaskName;} set
    {
        if(string.IsNullOrWhiteSpace(value))
        {
            throw new ArgumentException("Название задачи не может быть пустым");
        }
        else{
            _TaskName = value;
        }
    }
    }

    public int Priority {get {return _Priority;} set
    {
        if(value >= 1 && value <= 5)
        {
            _Priority = value;
        }
        else{
            throw new ArgumentException("Приоритет должен быть от 1 до 5");
        }
    }}

    public string Assignee
    {
        get {return _assignee;}
        set {
            if(string.IsNullOrWhiteSpace(value))
            {
                _assignee = "not assigned";
            }
        }
    }
    
    // Конструктор
    public Task(int taskId, string taskName, int priority, string assignee = "not assigned")
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        Assignee = assignee;
    }

    // Метод для отметки задачи как выполненной
    public virtual void MarkAsComplete()
    {
        isCompleted = true;
    }

    // Метод для получения деталей задачи
    public virtual void GetTaskDetails()
    {
        if (!string.IsNullOrWhiteSpace(About))
        {
            Console.WriteLine($"ID: {TaskId}, Название: {TaskName}, Приоритет: {Priority}\nОписание задачи: {About}");
        }
        else
        {
            Console.WriteLine($"ID: {TaskId}, Название: {TaskName}, Приоритет: {Priority}");
        }
    }

    // Метод для переназначения задачи
    public virtual void ReassignTo(string newAssignee)
    {
        Console.WriteLine($"Задача {TaskName} переназначена на {newAssignee}");
    }

    public virtual void Info()
    {
        Console.WriteLine($"Краткое описание задачи: {About}");
    }

    public virtual void Info(int about)
    {
        Console.WriteLine($"Задача выполнялась {about} часов");
    }

    public virtual void Info(bool about)
    {
        if (about)
        {
            Console.WriteLine($"Задача выполнена вовремя");
        }
        else
        {
            Console.WriteLine($"Задача не выполнена вовремя");
        }
        
    }
}
public class DelegateTask : Task
{
    // Дополнительный атрибут
    public DateTime DueDate { get; set; }
    
    public DelegateTask(int taskId, string taskName, int priority, DateTime dueDate) 
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
    }

    // Переопределенный метод для отметки задачи как выполненной
    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Задача {TaskName} отчмечена как выполненная.\nДата выполнения: {DueDate}");
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"Дата выполнения: {DueDate}");
    }

    public override void Info()
    {
        base.Info();
    }

    public override void Info(int about)
    {
        base.Info(about);
    }

    public override void Info(bool about)
    {
        base.Info(about);
    }
}
public class TeamTask : Task
{
    // Дополнительный атрибут
    public string TeamName { get; set; }
    
    // Конструктор
    public TeamTask(int taskId, string taskName, int priority, string teamName) 
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
    }

    // Переопределенный метод для переназначения задачи
    public override void ReassignTo(string newAssignee)
    {
        base.ReassignTo(newAssignee);
        Console.WriteLine($"Новая команда: {TeamName}");
    }

    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Команда {TeamName} выполнила задачу {TaskName}");
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"Команда, выполняющая задачу: {TeamName}");
    }

    public override void Info()
    {
        base.Info();
    }

    public override void Info(int about)
    {
        base.Info(about);
    }

    public override void Info(bool about)
    {
        base.Info(about);
    }
}
public class ResearchTask : Task
{
    // Дополнительный атрибут
    public string DataSource { get; set; }
    
    // Конструктор
    public ResearchTask(int taskId, string taskName, int priority, string dataSource) 
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
    }

    // Переопределенный метод для получения деталей задачи
    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"Источник данных: {DataSource}");
    }

    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Исследовательская задача '{TaskName}' выполнена");
    }

    public override void Info()
    {
        base.Info();
    }

    public override void Info(int about)
    {
        base.Info(about);
    }

    public override void Info(bool about)
    {
        base.Info(about);
    }
}

// Generic class
public class TaskCollection<T> where T : Task
{
    private List<T> _tasks = new List<T>();

    public void Add(T task)
    {
        _tasks.Add(task);
    }

    public void Remove(T task)
    {
        _tasks.Remove(task);
    }

    public void DisplayTasksAbout()
    {
        foreach (var tsk in _tasks)
        {
            tsk.GetTaskDetails();
        }
    }
}

// Создаем различные типы задач
        Console.WriteLine("1. Создание базовой задачи:");
        Task basicTask = new Task(1, "Базовая задача", 2);
        basicTask.About = "Пример обычной базовой задачи";
        basicTask.GetTaskDetails();
        
        Console.WriteLine("\n2. Создание делегированной задачи:");
        DelegateTask delegateTask = new DelegateTask(2, "Делегированная задача", 1, DateTime.Now.AddDays(3));
        delegateTask.About = "Пример делегированной задачи, в которой указывается дата";
        delegateTask.GetTaskDetails();

        Console.WriteLine("\n3. Создание командной задачи:");
        TeamTask teamTask = new TeamTask(3, "Командная задача", 3, "Команда разработки");
        teamTask.About = "Пример командной задачи, где указывается команда выполнения";
        teamTask.GetTaskDetails();

        Console.WriteLine("\n4. Создание исследовательской задачи:");
        ResearchTask researchTask = new ResearchTask(4, "Исследовательская задача", 2, "Google Scholar");
        researchTask.About = "Пример исследовательской информации где указывается источник данных";
        researchTask.GetTaskDetails();

        // // Отметка задач как выполненных
        // Console.WriteLine("\nОтмечаем задачи как выполненные:");
        // basicTask.MarkAsComplete();
        // delegateTask.MarkAsComplete();
        // teamTask.MarkAsComplete();
        // researchTask.MarkAsComplete();

        // // Переназначение задач
        // Console.WriteLine("\nПереназначаем задачи:");
        // basicTask.ReassignTo("Person1");
        // teamTask.ReassignTo("Person2");

        // // Получение обновленных деталей
        // Console.WriteLine("\nОбновленные детали задач:");
        // basicTask.GetTaskDetails();
        // delegateTask.GetTaskDetails();
        // teamTask.GetTaskDetails();
        // researchTask.GetTaskDetails();

Console.WriteLine();
basicTask.Info();
teamTask.Info(47);
researchTask.Info(false);
Console.WriteLine();

// Пример использования Generic класса, вывод всех описаний задач
TaskCollection<Task> taskss = new TaskCollection<Task>();
taskss.Add(basicTask);
taskss.Add(delegateTask); 
taskss.Add(teamTask);
taskss.Add(researchTask);

taskss.DisplayTasksAbout();
```
