```
public class Task
{
    // Базовые атрибуты
    protected int _TaskId;
    protected string _TaskName;
    protected int _Priority;
    private string _assignee;
    private bool isCompleted = false;

    public int TaskId {get {return _TaskId;} set
    {
        if(value > 0)
        {
            _TaskId = value;
        }
        else{
            throw new ArgumentException("ID задачи должен быть положительным числом");
        }
    }
    }

    public string TaskName {get {return _TaskName;} set
    {
        if(string.IsNullOrWhiteSpace(value))
        {
            throw new ArgumentException("Название задачи не может быть пустым");
        }
        else{
            _TaskName = value;
        }
    }
    }

    public int Priority {get {return _Priority;} set
    {
        if(value >= 1 && value <= 5)
        {
            _Priority = value;
        }
        else{
            throw new ArgumentException("Приоритет должен быть от 1 до 5");
        }
    }}

    public string Assignee
    {
        get {return _assignee;}
        set {
            if(string.IsNullOrWhiteSpace(value))
            {
                _assignee = "not assigned";
            }
        }
    }
    
    // Конструктор
    public Task(int taskId, string taskName, int priority, string assignee = "not assigned")
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        Assignee = assignee;
    }

    // Метод для отметки задачи как выполненной
    public virtual void MarkAsComplete()
    {
        isCompleted = true;
        Console.WriteLine($"Задача {TaskName} отмечена как выполненная");
    }

    // Метод для получения деталей задачи
    public virtual string GetTaskDetails()
    {
        return $"ID: {TaskId}, Название: {TaskName}, Приоритет: {Priority}";
    }

    // Метод для переназначения задачи
    public virtual void ReassignTo(string newAssignee)
    {
        Console.WriteLine($"Задача {TaskName} переназначена на {newAssignee}");
    }
}
public class DelegateTask : Task
{
    // Дополнительный атрибут
    public DateTime DueDate { get; set; }
    
    public DelegateTask(int taskId, string taskName, int priority, DateTime dueDate) 
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
    }

    // Переопределенный метод для отметки задачи как выполненной
    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Дата выполнения: {DueDate}");
    }
}
public class TeamTask : DelegateTask
{
    // Дополнительный атрибут
    public string TeamName { get; set; }
    
    // Конструктор
    public TeamTask(int taskId, string taskName, int priority, DateTime dueDate, string teamName) 
        : base(taskId, taskName, priority, dueDate)
    {
        TeamName = teamName;
    }

    // Переопределенный метод для переназначения задачи
    public override void ReassignTo(string newAssignee)
    {
        base.ReassignTo(newAssignee);
        Console.WriteLine($"Новая команда: {TeamName}");
    }
}

public interface IResearchTask
{
    string DataSource {get; set;}
}

public abstract class Task1
{
    // Базовые атрибуты
    protected int _TaskId1;
    protected string _TaskName1;
    protected int _Priority1;
    private string _assignee1;
    protected bool isCompleted1 = false;

    public int TaskId1 {get {return _TaskId1;} set
    {
        if(value > 0)
        {
            _TaskId1 = value;
        }
        else{
            throw new ArgumentException("ID задачи должен быть положительным числом");
        }
    }
    }

    public string TaskName1 {get {return _TaskName1;} set
    {
        if(string.IsNullOrWhiteSpace(value))
        {
            throw new ArgumentException("Название задачи не может быть пустым");
        }
        else{
            _TaskName1 = value;
        }
    }
    }

    public int Priority1 {get {return _Priority1;} set
    {
        if(value >= 1 && value <= 5)
        {
            _Priority1 = value;
        }
        else{
            throw new ArgumentException("Приоритет должен быть от 1 до 5");
        }
    }}

    public string Assignee1
    {
        get {return _assignee1;}
        set {
            if(string.IsNullOrWhiteSpace(value))
            {
                _assignee1 = "not assigned";
            }
        }
    }
    
    // Конструктор
    protected Task1(int taskId, string taskName, int priority, string assignee = "not assigned")
    {
        TaskId1 = taskId;
        TaskName1 = taskName;
        Priority1 = priority;
        Assignee1 = assignee;
    }

    // Метод для отметки задачи как выполненной
    public virtual void MarkAsComplete1()
    {
        isCompleted1 = true;
        Console.WriteLine($"Задача {TaskName1} отмечена как выполненная");
    }

    // Метод для получения деталей задачи
    protected virtual string GetTaskDetails1()
    {
        return $"ID: {TaskId1}, Название: {TaskName1}, Приоритет: {Priority1}";
    }

    // Метод для переназначения задачи
    public virtual void ReassignTo(string newAssignee)
    {
        Console.WriteLine($"Задача {TaskName1} переназначена на {newAssignee}");
    }
}

public class ResearchTask : Task1, IResearchTask
{
    // Дополнительный атрибут
    public string DataSource { get; set; }
    
    // Конструктор
    public ResearchTask(int taskId, string taskName, int priority, string dataSource) 
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
    }

    // Переопределенный метод для получения деталей задачи
    public void GetTaskDetails2()
    {
        Console.WriteLine(base.GetTaskDetails1() + $", Источник данных: {DataSource}") ;
    }

    public virtual void MarkAsComplete2()
    {
        isCompleted1 = true;
        Console.WriteLine($"Задача {TaskName1} отмечена как выполненная");
    }
}

// Создаем различные типы задач
        Console.WriteLine("1. Создание базовой задачи:");
        Task basicTask = new Task(1, "Базовая задача", 2);
        Console.WriteLine(basicTask.GetTaskDetails());
        
        Console.WriteLine("\n2. Создание делегированной задачи:");
        DelegateTask delegateTask = new DelegateTask(2, "Делегированная задача", 1, DateTime.Now.AddDays(3));
        Console.WriteLine(delegateTask.GetTaskDetails());

        Console.WriteLine("\n3. Создание командной задачи:");
        TeamTask teamTask = new TeamTask(3, "Командная задача", 3, DateTime.Now.AddDays(3), "Команда разработки");
        Console.WriteLine(teamTask.GetTaskDetails());

        Console.WriteLine("\n4. Создание исследовательской задачи:");
        ResearchTask researchTask = new ResearchTask(4, "Исследовательская задача", 2, "Google Scholar");
        researchTask.GetTaskDetails2();

        // Отметка задач как выполненных
        Console.WriteLine("Отмечаем задачи как выполненные:");
        basicTask.MarkAsComplete();
        delegateTask.MarkAsComplete();
        teamTask.MarkAsComplete();
        researchTask.MarkAsComplete2();

        // Переназначение задач
        Console.WriteLine("\nПереназначаем задачи:");
        basicTask.ReassignTo("Person1");
        teamTask.ReassignTo("Person2");

        // Получение обновленных деталей
        Console.WriteLine("\nОбновленные детали задач:");
        Console.WriteLine(basicTask.GetTaskDetails());
        Console.WriteLine(delegateTask.GetTaskDetails());
        Console.WriteLine(teamTask.GetTaskDetails());
        researchTask.GetTaskDetails2();
```
