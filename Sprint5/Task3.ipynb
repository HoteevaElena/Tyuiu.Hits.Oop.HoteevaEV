<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Task в C#, который будет представлять задачи внутри
проекта. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Task:

• Атрибуты: ID задачи (TaskId), Название задачи (TaskName), Приоритет
задачи (Priority).

• Методы:
o MarkAsComplete(): метод для отметки задачи как выполненной.
o GetTaskDetails(): метод для получения деталей задачи.
o ReassignTo(): метод для переназначения задачи другому члену
команды.

Требования к производным классам:
1. ДелегатскаяЗадача (DelegateTask): Должна содержать дополнительные
атрибуты, такие как Дата выполнения (DueDate).
Метод MarkAsComplete() должен быть переопределен для включения даты
выполнения в сообщение о завершении задачи.
2. КоманднаяЗадача (TeamTask): Должна содержать дополнительные атрибуты,
такие как Команда (TeamName). Метод ReassignTo() должен быть
переопределен для указания нового члена команды, которому будет
переназначена задача.
3. ИсследовательскаяЗадача (ResearchTask) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Исходные данные
(DataSource). Метод GetTaskDetails() должен быть переопределен для
отображения источников данных, используемых в задаче, вместе с другими
деталями задачи.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.


<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

// Делегат для уведомлений о задачах
public delegate void TaskNotificationHandler(string message);

public class Task
{
    // Базовые атрибуты
    public int TaskId;
    public string TaskName;
    public int Priority;
    private bool isCompleted = false;
    public DateTime CreatedDate { get; private set; }
    public string Description { get; set; }
    public string Category { get; set; }
    public int EstimatedHours { get; set; }

    // Событие для уведомлений
    public event TaskNotificationHandler OnTaskStatusChanged;

    public Task(int taskId, string taskName, int priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        CreatedDate = DateTime.Now;
        Description = "Без описания";
        Category = "Общая";
        EstimatedHours = 1;
    }

    // Метод для вызова события
    protected virtual void RaiseStatusChanged(string message)
    {
        OnTaskStatusChanged?.Invoke(message);
    }

    // Метод для отметки задачи как выполненной
    public virtual void MarkAsComplete()
    {
        isCompleted = true;
        string message = $"Задача {TaskName} отмечена как выполненная";
        Console.WriteLine(message);
        RaiseStatusChanged(message);
    }

    // Метод для получения деталей задачи
    public virtual string GetTaskDetails()
    {
        return $"ID: {TaskId}, Название: {TaskName}, Приоритет: {Priority}, " +
               $"Категория: {Category}, Часы: {EstimatedHours}";
    }

    // Метод для переназначения задачи
    public virtual void ReassignTo(string newAssignee)
    {
        string message = $"Задача {TaskName} переназначена на {newAssignee}";
        Console.WriteLine(message);
        RaiseStatusChanged(message);
    }

    // Новый метод Изменить приоритет
    public virtual void ChangePriority(int newPriority)
    {
        Priority = newPriority;
        string message = $"Приоритет задачи {TaskName} изменен на {newPriority}";
        Console.WriteLine(message);
        RaiseStatusChanged(message);
    }

    // Новый метод: обновить описание
    public virtual void UpdateDescription(string newDescription)
    {
        Description = newDescription;
        Console.WriteLine($"Описание задачи {TaskName} обновлено");
    }

    // Новый метод: проверить статус
    public virtual void CheckStatus()
    {
        string status = isCompleted ? "выполнена" : "в процессе";
        Console.WriteLine($"Задача {TaskName}: {status}");
    }
}

public class DelegateTask : Task
{
    // Дополнительные атрибуты
    public DateTime DueDate;
    public string Assignee { get; set; }
    public string Department { get; set; }
    public bool IsUrgent { get; set; }

    public DelegateTask(int taskId, string taskName, int priority, DateTime dueDate) 
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
        Assignee = "Не назначен";
        Department = "Общий";
        IsUrgent = false;
    }

    // Переопределенный метод для отметки задачи как выполненной
    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Дата выполнения: {DueDate}, Исполнитель: {Assignee}");
    }

    // Новый метод Установить исполнителя
    public void SetAssignee(string assignee)
    {
        Assignee = assignee;
        Console.WriteLine($"Исполнитель задачи {TaskName} установлен: {assignee}");
        RaiseStatusChanged($"Задача {TaskName} назначена на {assignee}");
    }

    // Новый метод Проверить срок
    public void CheckDueDate()
    {
        TimeSpan timeLeft = DueDate - DateTime.Now;
        if (timeLeft.TotalDays < 0)
            Console.WriteLine($"Срок задачи {TaskName} истек!");
        else
            Console.WriteLine($"До срока задачи {TaskName} осталось: {timeLeft.Days} дней");
    }
}

public class TeamTask : Task
{
    // Дополнительные атрибуты
    public string TeamName;
    public int TeamSize { get; set; }
    public string ProjectName { get; set; }
    public List<string> TeamMembers { get; private set; }
    
    public TeamTask(int taskId, string taskName, int priority, string teamName) 
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
        TeamSize = 1;
        ProjectName = "Основной проект";
        TeamMembers = new List<string>();
    }

    // Переопределенный метод для переназначения задачи
    public override void ReassignTo(string newAssignee)
    {
        base.ReassignTo(newAssignee);
        Console.WriteLine($"Новая команда: {TeamName}");
    }

    // Новый метод Добавить участника команды
    public void AddTeamMember(string memberName)
    {
        TeamMembers.Add(memberName);
        TeamSize = TeamMembers.Count;
        Console.WriteLine($"Участник {memberName} добавлен в команду {TeamName}");
    }

    // Новый метод Показать всех участников
    public void ShowTeamMembers()
    {
        Console.WriteLine($"Участники команды {TeamName}:");
        foreach (var member in TeamMembers)
        {
            Console.WriteLine($" - {member}");
        }
    }

    // Новый метод Изменить проект
    public void ChangeProject(string newProject)
    {
        ProjectName = newProject;
        Console.WriteLine($"Проект команды {TeamName} изменен на: {newProject}");
    }
}

public class ResearchTask : Task
{
    // Дополнительные атрибуты
    public string DataSource;
    public string ResearchTopic { get; set; }
    public int PagesCount { get; set; }
    public List<string> References { get; private set; }
    
    public ResearchTask(int taskId, string taskName, int priority, string dataSource) 
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
        ResearchTopic = "Общая тема";
        PagesCount = 1;
        References = new List<string>();
    }

    // Переопределенный метод для получения деталей задачи
    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Источник данных: {DataSource}, Тема: {ResearchTopic}";
    }

    // Новый метод Добавить ссылку
    public void AddReference(string reference)
    {
        References.Add(reference);
        Console.WriteLine($"Ссылка добавлена в исследование {TaskName}");
    }

    // Новый метод Показать все ссылки
    public void ShowReferences()
    {
        Console.WriteLine($"Ссылки для исследования {TaskName}:");
        foreach (var reference in References)
        {
            Console.WriteLine($" - {reference}");
        }
    }

    // Новый метод Обновить тему исследования
    public void UpdateResearchTopic(string newTopic)
    {
        ResearchTopic = newTopic;
        Console.WriteLine($"Тема исследования {TaskName} обновлена: {newTopic}");
    }
}

// Менеджер задач для работы с коллекциями
public class TaskManager
{
    private List<Task> tasks;
    
    // Событие для уведомлений менеджера
    public event TaskNotificationHandler OnManagerNotification;

    public TaskManager()
    {
        tasks = new List<Task>();
    }

    // Добавить задачу
    public void AddTask(Task task)
    {
        tasks.Add(task);
        // Подписываемся на событие задачи
        task.OnTaskStatusChanged += TaskStatusChangedHandler;
        OnManagerNotification?.Invoke($"Задача '{task.TaskName}' добавлена в менеджер");
    }

    // Обработчик изменения статуса задачи
    private void TaskStatusChangedHandler(string message)
    {
        Console.WriteLine($"[Менеджер] {message}");
    }

    // Показать все задачи
    public void ShowAllTasks()
    {
        Console.WriteLine("\n=== ВСЕ ЗАДАЧИ ===");
        foreach (var task in tasks)
        {
            Console.WriteLine(task.GetTaskDetails());
        }
    }

    // Найти задачи по имени
    public void FindTasksByName(string name)
    {
        var foundTasks = tasks.FindAll(t => t.TaskName.Contains(name));
        Console.WriteLine($"\n=== НАЙДЕННЫЕ ЗАДАЧИ ПО '{name}' ===");
        foreach (var task in foundTasks)
        {
            Console.WriteLine(task.GetTaskDetails());
        }
    }

    // Показать задачи по приоритету
    public void ShowTasksByPriority(int priority)
    {
        var priorityTasks = tasks.FindAll(t => t.Priority == priority);
        Console.WriteLine($"\n=== ЗАДАЧИ С ПРИОРИТЕТОМ {priority} ===");
        foreach (var task in priorityTasks)
        {
            Console.WriteLine(task.GetTaskDetails());
        }
    }
}

static void NotificationHandler(string message)
    {
        Console.WriteLine($"[УВЕДОМЛЕНИЕ] {message}");
    }

    
// Создаем менеджер задач
TaskManager taskManager = new TaskManager();
taskManager.OnManagerNotification += NotificationHandler;

Console.WriteLine("1. Создание базовой задачи:");
Task basicTask = new Task(1, "Базовая задача", 2);
basicTask.OnTaskStatusChanged += NotificationHandler;
taskManager.AddTask(basicTask);

Console.WriteLine("\n2. Создание делегированной задачи:");
DelegateTask delegateTask = new DelegateTask(2, "Делегированная задача", 1, DateTime.Now.AddDays(3));
delegateTask.OnTaskStatusChanged += NotificationHandler;
delegateTask.SetAssignee("Иван Иванов");
delegateTask.CheckDueDate();
taskManager.AddTask(delegateTask);

Console.WriteLine("\n3. Создание командной задачи:");
TeamTask teamTask = new TeamTask(3, "Командная задача", 3, "Команда разработки");
teamTask.OnTaskStatusChanged += NotificationHandler;
teamTask.AddTeamMember("Петр Петров");
teamTask.AddTeamMember("Мария Сидорова");
teamTask.ShowTeamMembers();
taskManager.AddTask(teamTask);

Console.WriteLine("\n4. Создание исследовательской задачи:");
ResearchTask researchTask = new ResearchTask(4, "Исследовательская задача", 2, "Google Scholar");
researchTask.OnTaskStatusChanged += NotificationHandler;
researchTask.AddReference("Источник 1");
researchTask.AddReference("Источник 2");
researchTask.ShowReferences();
taskManager.AddTask(researchTask);

// Демонстрация работы с задачами
Console.WriteLine("\n=== РАБОТА С ЗАДАЧАМИ ===");

// Отметка задач как выполненных
Console.WriteLine("Отмечаем задачи как выполненные:");
basicTask.MarkAsComplete();
delegateTask.MarkAsComplete();

// Переназначение задач
Console.WriteLine("\nПереназначаем задачи:");
basicTask.ReassignTo("Новый исполнитель");
teamTask.ReassignTo("Новая команда");

// Использование новых методов
Console.WriteLine("\n=== ДОПОЛНИТЕЛЬНЫЕ МЕТОДЫ ===");
basicTask.ChangePriority(5);
basicTask.CheckStatus();

teamTask.ChangeProject("Новый проект");

researchTask.UpdateResearchTopic("Искусственный интеллект");

// Работа с коллекциями через менеджер
Console.WriteLine("\n=== РАБОТА МЕНЕДЖЕРА ЗАДАЧ ===");
taskManager.ShowAllTasks();
taskManager.FindTasksByName("задача");
taskManager.ShowTasksByPriority(2);

// Добавим еще одну задачу для демонстрации
Console.WriteLine("\n5. Добавляем дополнительную задачу:");
Task extraTask = new Task(5, "Дополнительная задача", 1);
extraTask.OnTaskStatusChanged += NotificationHandler;
taskManager.AddTask(extraTask);
extraTask.MarkAsComplete();

taskManager.ShowAllTasks();
```
